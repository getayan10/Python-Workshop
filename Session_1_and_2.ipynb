#NUMPY

```python
import numpy as np
l1=[1,2,3,4.0]
l2=[[1,2,3,],[4,5,6],[7,8,9]]
```

```python
a1=np.array(l1)
print("L1:",l1)
print(type(l1))
print("Array 1:",a1,"\n",type(a1))

print(l1[0],type(l1[0]))
print(l1[-1],type(l1[-1]))
print(a1[0],type(a1[0]))
```

```python
a2=np.matrix(l2)
print(a2)
```

```python
print(a2.shape)
print(a2.sum())
print(np.linalg.det(a2))
```

```python
import pandas as pd
p1=pd.Series(l1,index=["a","b","c","d"])
print(p1)
print(type(p1))
```
