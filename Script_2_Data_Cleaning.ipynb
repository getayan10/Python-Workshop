```python
df2=df.copy()
df2=df2.dropna(axis=1)
print(df2.shape)
print(df2.isnull().sum())
```

```python
df3=df.copy()
x=np.mean(df3.bmi)
print(x)
print(df3.bmi.isnull().sum())
df3.bmi=df3.bmi.fillna(x)
print(df3.bmi.isnull().sum())
```

```python
x=np.mean(df3.age)
print(x)
print(np.round(x,0))
print(df3.age.isnull().sum())
df3.age=df3.age.fillna(np.round(x,0))
print(df3.age.isnull().sum())
```
