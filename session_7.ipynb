```python
import pandas as pd
import numpy as np
df=pd.read_excel("/content/drive/MyDrive/Dataset/Session 5/Hotels.xlsx")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(9,4))
plt.hist(df['Revenue'], bins=32, color='green',linewidth=2,edgecolor='red')
plt.title('Distribution of Hotel Revenue', fontsize=16)
plt.xlabel('Revenue')
plt.ylabel('Number of Hotels')
plt.show()
```

```python
plt.figure(figsize=(9,4))
sns.histplot(df['Revenue'],kde=True, bins=32, color='green',linewidth=2,edgecolor='red')
plt.title('Distribution of Hotel Revenue', fontsize=16)
plt.xlabel('Revenue')
plt.ylabel('Number of Hotels')
plt.show()
```

```python
plt.figure(figsize=(10,3))
#plt.boxplot(df['Revenue'],orientation='horizontal')
sns.boxplot(x=df['Revenue'],width=0.2)
plt.title('Distribution of Hotel Revenue', fontsize=16)
plt.xlabel('Revenue')
plt.show()
```

```python
plt.figure(figsize=(12,4))
sns.countplot(y='Location',data=df, order=df['Location'].value_counts().index)
plt.title('Number of Hotels per Location', fontsize=16)
plt.xlabel('Number of Hotels')
plt.ylabel('Location')
plt.show()
```

```python
data=df['Location'].value_counts()
print(data.index)
print(data.values)
```

```python
from matplotlib.lines import lineStyles
plt.figure(figsize=(12,3))
plt.bar(x=df['Location'].value_counts().index,height=df['Location'].value_counts().values)
plt.title('Number of Hotels per Location', fontsize=16)
plt.xlabel('Number of Hotels')
plt.grid(True, linestyle='-',alpha=0.4)
plt.xticks(rotation=90)
plt.ylabel("Location")
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.regplot(x='Number of Rooms',y='Revenue',data=df, scatter_kws={'alpha':0.8}, line_kws={'color':'red'})
plt.title('Revenue VS Number of Rooms', fontsize=16)
plt.xlabel('Number of Rooms')

plt.ylabel("Revenue")
plt.show()
```

```python
contigency_table=pd.crosstab(df['Location'],df['Occupancy Rate'])
print(contigency_table)
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(contigency_table, annot=True,fmt='f',cmap='cubehelix')
plt.title('Occupancy Rate per Location', fontsize=16)
plt.xlabel('Occupancy Rate')
plt.ylabel("Location")
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.boxplot(x='Revenue', y='Location', data=df)
plt.title('Occupancy Rate per Location', fontsize=16)
plt.xlabel('Occupancy Rate')
plt.ylabel("Location")
plt.show()
```
